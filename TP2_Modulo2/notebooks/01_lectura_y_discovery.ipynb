## 6. Distribución del target

```python
print('Distribución del target (smoking):')
print(df['smoking'].value_counts())
print()
print(df['smoking'].value_counts(normalize=True) * 100)
```

```python
plt.figure(figsize=(6, 5))
conteo = df['smoking'].value_counts()
plt.pie(conteo, labels=['No fumador (0)', 'Fumador (1)'], autopct='%1.1f%%',
        startangle=90, colors=['#2ecc71', '#e74c3c'])
plt.title('Distribución de la variable target: smoking')
plt.tight_layout()
plt.show()
```

## 7. Conclusiones del Discovery

- Dataset etiquetado: **50.000 filas**, 26 features + 1 target (`smoking`)
- Dataset sin etiquetar: **5.692 filas**, 26 features (sin target)
- **Sin valores faltantes** en ningún dataset
- Variables categóricas: `gender` (M/F), `oral` (Y/N), `tartar` (Y/N)
- Target **desbalanceado**: ~63% no fumadores (0) vs ~37% fumadores (1)
- Columna `ID` es un identificador, no una feature predictiva → se descarta
- Todas las variables numéricas son continuas o discretas con sentido médico
