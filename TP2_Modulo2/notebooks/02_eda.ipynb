## Notebook 02 - Análisis Exploratorio de Datos (EDA)

**Objetivo:** Analizar en profundidad las distribuciones de las variables, relaciones con el target y detectar outliers.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
sns.set_theme(style='whitegrid')
```

```python
df = pd.read_excel('../data/raw/dataset_etiquetado.xlsx')
print(df.shape)
```

## 1. Variables categóricas vs target

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for ax, col in zip(axes, ['gender', 'oral', 'tartar']):
    sns.countplot(data=df, x=col, hue='smoking', ax=ax,
                  palette=['#2ecc71', '#e74c3c'])
    ax.set_title(f'{col} vs smoking')
    ax.legend(title='smoking', labels=['No fumador', 'Fumador'])
plt.tight_layout()
plt.show()
```

## 2. Variables numéricas - distribución vs target

```python
num_cols = ['age', 'height(cm)', 'weight(kg)', 'waist(cm)', 'hemoglobin',
            'Gtp', 'ALT', 'triglyceride', 'HDL', 'LDL', 'Cholesterol',
            'systolic', 'relaxation', 'fasting blood sugar']

fig, axes = plt.subplots(4, 4, figsize=(18, 14))
axes = axes.flatten()

for i, col in enumerate(num_cols):
    sns.histplot(data=df, x=col, hue='smoking', ax=axes[i],
                 kde=True, stat='density', common_norm=False,
                 palette=['#2ecc71', '#e74c3c'], alpha=0.6)
    axes[i].set_title(col)
    axes[i].legend(title='smoking', labels=['No fum.', 'Fum.'])

for j in range(len(num_cols), len(axes)):
    axes[j].set_visible(False)

plt.suptitle('Distribución de variables numéricas por clase', fontsize=14, y=1.01)
plt.tight_layout()
plt.show()
```

## 3. Boxplots para detectar outliers

```python
fig, axes = plt.subplots(5, 5, figsize=(20, 18))
axes = axes.flatten()
all_num = df.select_dtypes(include=[np.number]).drop(columns=['ID', 'smoking', 'dental caries']).columns.tolist()

for i, col in enumerate(all_num):
    sns.boxplot(data=df, x='smoking', y=col, ax=axes[i],
                palette=['#2ecc71', '#e74c3c'])
    axes[i].set_title(col)
    axes[i].set_xticklabels(['No fum.', 'Fum.'])

for j in range(len(all_num), len(axes)):
    axes[j].set_visible(False)

plt.suptitle('Boxplots por variable y clase', fontsize=14, y=1.01)
plt.tight_layout()
plt.show()
```

## 4. Mapa de correlaciones

```python
# Codificación temporal para calcular correlaciones
df_corr = df.copy()
df_corr['gender']  = df_corr['gender'].map({'M': 1, 'F': 0})
df_corr['oral']    = df_corr['oral'].map({'Y': 1, 'N': 0})
df_corr['tartar']  = df_corr['tartar'].map({'Y': 1, 'N': 0})
df_corr = df_corr.drop(columns=['ID'])

plt.figure(figsize=(14, 11))
corr = df_corr.corr(method='pearson')
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdYlGn',
            mask=mask, vmin=-1, vmax=1, linewidths=0.5)
plt.title('Mapa de correlaciones (Pearson)', fontsize=14)
plt.tight_layout()
plt.show()
```

## 5. Correlación con el target

```python
corr_target = corr['smoking'].drop('smoking').sort_values(ascending=False)

plt.figure(figsize=(10, 6))
colors = ['#e74c3c' if v > 0 else '#2ecc71' for v in corr_target.values]
plt.barh(corr_target.index, corr_target.values, color=colors)
plt.axvline(0, color='black', linewidth=0.8)
plt.title('Correlación de cada variable con el target (smoking)')
plt.xlabel('Correlación de Pearson')
plt.tight_layout()
plt.show()

print('Top 10 más correlacionadas:')
print(corr_target.head(10))
```

## 6. Análisis de variables más importantes

```python
# hemoglobin y gender son los más correlacionados
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Hemoglobina
sns.kdeplot(data=df, x='hemoglobin', hue='smoking',
            fill=True, alpha=0.4, palette=['#2ecc71', '#e74c3c'], ax=axes[0])
axes[0].set_title('Distribución de hemoglobina por clase')

# Género
ct = pd.crosstab(df['gender'], df['smoking'], normalize='index') * 100
ct.plot(kind='bar', ax=axes[1], color=['#2ecc71', '#e74c3c'], rot=0)
axes[1].set_title('% fumadores por género')
axes[1].set_ylabel('Porcentaje')
axes[1].legend(['No fumador', 'Fumador'])

plt.tight_layout()
plt.show()
```

## 7. Conclusiones del EDA

- **Variables más correlacionadas con smoking:** `hemoglobin`, `gender` (M), `height(cm)`, `weight(kg)`, `Gtp`, `ALT`, `waist(cm)`
- Los **hombres fuman significativamente más** que las mujeres
- Los **fumadores tienen mayor hemoglobina**, mayor peso/altura y cintura
- Existen outliers en variables como `Gtp`, `ALT`, `triglyceride`, `serum creatinine` → se manejarán con transformaciones log o winsorización
- La variable `Urine protein` tiene muy pocos valores distintos → se puede tratar como categórica ordinal
- `hearing(left)` y `hearing(right)` tienen valores muy concentrados → baja varianza
